```python
def attrition_rate_grouping(df, group_column):
    df_temp = df.groupby(group_column).agg({'Attrition_Num': 'sum', 'EmployeeNumber': 'count'})
    df_temp.rename(columns={'EmployeeNumber': 'EmployeeCount'}, inplace=True)
    df_temp['Attrition_Rate'] = (df_temp['Attrition_Num'] / df_temp['EmployeeCount'] * 100).round(2)
    df_temp = df_temp.reset_index()
    plt.figure(figsize=(12, 4))
    ax = sns.barplot(data=df_temp, x=group_column, y='Attrition_Rate')
    for container in ax.containers:
        ax.bar_label(container, fmt='%.2f%%')
    plt.show()
    plt.close()
    return df_temp
```

```python
income_bins = [1000, 3000, 5000, 7000, 10000, 20000]

income_labels = [
    "\\$1K–$3K",
    "\\$3K–$5K",
    "\\$5K–$7K",
    "\\$7K–$10K",
    "\\$10K+"
]

income_groups = pd.cut(
    df["MonthlyIncome"],
    bins=income_bins,
    labels=income_labels,
    include_lowest=True
)
```

```python
income_analysis = (
    df.groupby(income_groups, observed=True)
      .agg(
          Employees=("Attrition_Num", "size"),
          Attritions=("Attrition_Num", "sum"),
          Attrition_Rate=("Attrition_Num", "mean")
      )
)

income_analysis
```
