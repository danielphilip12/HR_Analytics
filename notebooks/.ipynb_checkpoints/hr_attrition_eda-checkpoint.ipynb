```python
def attrition_rate_grouping(df, group_column):
    df_temp = df.groupby(group_column).agg({'Attrition_Num': 'sum', 'EmployeeNumber': 'count'})
    df_temp.rename(columns={'EmployeeNumber': 'EmployeeCount'}, inplace=True)
    df_temp['Attrition_Rate'] = (df_temp['Attrition_Num'] / df_temp['EmployeeCount'] * 100).round(2)
    df_temp = df_temp.reset_index()
    plt.figure(figsize=(12, 4))
    ax = sns.barplot(data=df_temp, x=group_column, y='Attrition_Rate')
    for container in ax.containers:
        ax.bar_label(container, fmt='%.2f%%')
    plt.show()
    plt.close()
    return df_temp
```

### Finding

Attrition is highest among employees with 0–5 years under their current manager (18.87%) and declines among employees with longer manager tenure. The 11+ year group has a low attrition rate of 4.11%, although it represents a relatively small employee population.

## Years at company vs Age

```python
sns.scatterplot(data=df, x='YearsAtCompany', y='Age')
```

```python
df[['YearsAtCompany', 'Age']].describe()
```

```python
df_years_vs_age = df[['YearsAtCompany', 'Age', 'Attrition_Num']]
```

```python
age_bins = [18, 25, 35, 45, 55, np.inf]

age_labels = ["18-25", '26-35', '36-45', '46-55', '55+']

df_years_vs_age['AgeGroup'] = pd.cut(df_years_vs_age['Age'], bins=age_bins, labels=age_labels, include_lowest=True)

df_years_vs_age.head()
```

```python
yac_bins = [0, 5, 10, np.inf]

yac_labels = [
    '0-5 Years',
    '6-10 Years',
    '11+ Years'
]

df_years_vs_age['YearsAtCompanyGroup'] = pd.cut(df_years_vs_age['YearsAtCompany'], bins=yac_bins, labels=yac_labels, include_lowest=True)

df_years_vs_age.head()
```

```python
pd.pivot_table(data=df_years_vs_age,
               index='YearsAtCompanyGroup',
               columns='AgeGroup', 
               values='Attrition_Num', aggfunc='mean')
```

```python
pd.pivot_table(data=df_years_vs_age,
               index='YearsAtCompanyGroup',
               columns='AgeGroup', 
               values='Attrition_Num', aggfunc='count')
```

> Attrition is particularly high among younger employees with shorter company tenure. Among employees ages 26–35, attrition falls from 24.69% for those with 0–5 years at the company to 9.38% for those with 11+ years.
